# view3 check
Run all, then page through each table in your notebook's dark and light themes.

```
import polars as pl
import view3

penguins = pl.read_csv("https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv", null_values="NA")
```

## Every distinct value, paged

```
view3.view(penguins.select("body_mass_g").unique().sort("body_mass_g"));
```

## The same as a scroll box (`pages=False`)

```
view3.view(penguins.select("body_mass_g").unique().sort("body_mass_g"), pages=False, height=300);
```

## Wide: column pages

```
view3.view(penguins.with_columns(
    bill_ratio=pl.col("bill_length_mm") / pl.col("bill_depth_mm"),
    mass_kg=pl.col("body_mass_g") / 1000,
    label=pl.concat_str("species", "island", "sex", separator=" / "),
    note=pl.lit("a long note that goes past forty characters, cut with a hover"),
));
```

## Row limit, bigger pages, empty

```
view3.view(penguins, n=100, page_size=25);
```

```
view3.view(penguins.filter(pl.col("body_mass_g") > 99999));
```

## Viewing returns the data unchanged

```
before = penguins.clone()
view3.view(penguins).equals(before)
```

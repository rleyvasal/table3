# table3 check
Run all, then page through each table in your notebook's dark and light themes.

```
import polars as pl
import table3

penguins = pl.read_csv("https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv", null_values="NA")
```

## Every distinct value, paged

```
table3.show(penguins.select("body_mass_g").unique().sort("body_mass_g"));
```

## The same as a scroll box (`pages=False`)

```
table3.show(penguins.select("body_mass_g").unique().sort("body_mass_g"), pages=False, height=300);
```

## Wide: column pages

```
table3.show(penguins.with_columns(
    bill_ratio=pl.col("bill_length_mm") / pl.col("bill_depth_mm"),
    mass_kg=pl.col("body_mass_g") / 1000,
    label=pl.concat_str("species", "island", "sex", separator=" / "),
    note=pl.lit("a long note that goes past forty characters, cut with a hover"),
));
```

## Row limit, bigger pages, empty

```
table3.show(penguins, n=100, page_size=25);
```

```
table3.show(penguins.filter(pl.col("body_mass_g") > 99999));
```

## Explorer: search, click a header to sort, filter under each column

```
dated = penguins.with_columns(
    day=pl.date(2007, 1, 1) + pl.duration(days=pl.int_range(pl.len()) * 3),
)
table3.view(dated);
```

```
table3.view(pl.DataFrame({"x": range(10_000)}), height=300);
```

## Live explorer (pip install table3[live]): every row, through the kernel
Search, sort and filter cover all 2,000,000 rows. Reopen the notebook without its kernel: the first 1,000 rows show instead, and the status line says so.

```
big = pl.DataFrame({"x": range(2_000_000)}).with_columns(
    s=pl.format("item {}", "x"), half=pl.col("x") / 2)
table3.view(big, n=1000);
```

## Viewing returns the data unchanged

```
before = penguins.clone()
table3.show(penguins).equals(before)
```
